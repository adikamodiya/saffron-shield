```python
# R4 task 1 — ingest
# The claims_daily job is triggered by FILE ARRIVAL on /Volumes/claims/raw/landing/claims_raw/.
# This task moves the arrived batch files into /Volumes/claims/raw/landing/claims/ —
# the folder the SDP pipeline's bronze.claims Auto Loader actually watches.
# Idempotent: safe to re-run (a no-op when nothing has arrived).

SRC = "/Volumes/claims/raw/landing/claims_raw"
DST = "/Volumes/claims/raw/landing/claims"

dbutils.fs.mkdirs(SRC)   # creates the trigger folder on the very first run
dbutils.fs.mkdirs(DST)

moved = []
for f in dbutils.fs.ls(SRC):
    if not f.path.endswith("/"):
        name = f.path.rstrip("/").split("/")[-1]
        dbutils.fs.mv(f.path, f"{DST}/{name}")
        moved.append(name)

print(f"Ingested {len(moved)} file(s): {moved}")
```
